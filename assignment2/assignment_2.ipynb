```python
import random
random.seed(1024170415)
```

```python
numbers=[]
for i in range(100):
 numbers.append(random.randint(100,900))
numbers
```

```python
oddcount=0
for i in numbers:
    if i%2!=0:
     oddcount+=1

oddcount  
```

```python
evencount=0
for i in numbers:
    if i%2==0:
      evencount+=1
evencount    
```

```python
def isprime(i):
    if i<2:
        return False

    for j in range(2,i):
        if i%j ==0:
            return False
    return True

primenumbers = [i for i in numbers if isprime(i)]
primenumbers
        
```

```python
len(primenumbers)
```

```python
most_frequent = numbers[0]
max_count = numbers.count(numbers[0])

for i in numbers:
    count = numbers.count(i)

    if count > max_count:
        max_count = count
        most_frequent = i

print("Most frequent number:", most_frequent)
print("Number of times it occurs:", max_count)
```

```python
digits= [1,0,2,4,1,7,0,4,1,5]
A= set()
for i in digits:
    A.add(i*7)

B=set()
for i in digits:
    B.add(i*9)

A
```

```python
B
```

```python
unionset=A.union(B)
unionset
```

```python
intersectionset=A.intersection(B)
intersectionset
```

```python
k=A.difference(B)
```

```python
k
```

```python
symmetric=A.symmetric_difference(B)
symmetric
```

```python
# difference() gives elements present in one set but not the other,
# while symmetric_difference() gives elements that are in either set but not in both.
```

```python
print(A.issubset(B))
```

```python
print(A.issuperset(B))
```
