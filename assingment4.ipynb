```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])
print("original array:", arr)

add_arr = arr + 2
print("after addition of 2:", add_arr)
mul_arr = arr * 3
print("after multiplying by 3:", mul_arr)
div_arr = arr / 2
print("after dividing by 2:", div_arr)
```

```python
arr = np.array([1, 2, 3, 6, 4, 5])
reversed_arr = arr[::-1]
print("original array:", arr)
print("reversed array:", reversed_arr)
```

```python
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
values, counts = np.unique(x, return_counts=True)
most_freq_x = values[np.argmax(counts)]
indices_x = np.where(x == most_freq_x)[0]
print("array x:", x)
print("frequent value:", most_freq_x)
print("indices:", indices_x)
```

```python
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])
values, counts = np.unique(y, return_counts=True)
most_freq_y = values[np.argmax(counts)]
indices_y = np.where(y == most_freq_y)[0]
print("array y:", y)
print("most frequent value:", most_freq_y)
print("indices:", indices_y)
```

```python
arr = np.array([[10, 20, 30], [40, 50, 60], [70, 80, 90]])
print("2-D array:\n", arr)

element1 = arr[0, 1]
print("1st Row, 2nd column element:", element1)

element2 = arr[2, 0]
print("3rd Row, 1st column element:", element2)
```

```python
Vijaykiran = np.linspace(10, 100, 25)
print("Array Vijaykiran:", Vijaykiran)

print("Dimensions:", Vijaykiran.ndim)
print("Shape:", Vijaykiran.shape)
print("Total Elements:", Vijaykiran.size)
print("Data Type:", Vijaykiran.dtype)
print("Total Bytes:", Vijaykiran.nbytes)

transpose_reshape = Vijaykiran.reshape(25, 1)
print("Transpose using reshape():\n", transpose_reshape)
```

Since `Vijaykiran` is a 1-D array, applying the `.T` attribute does not change it, as transpose has no effect on a 1-D array. It only works properly on arrays with 2 or more dimensions. To actually see a transpose-like effect using `.T`, the array first needs to be reshaped into a 2-D array.

```python
print("using T attribute on 1-D array (no change):", Vijaykiran.T.shape)

Vijaykiran_2d = Vijaykiran.reshape(25, 1)
print("using T attribute on reshaped 2-D array:\n", Vijaykiran_2d.T)
```
