```python
steps = np.array([ [5000, 6200, 7100], [8000, 7500, 9000], [4500, 5100, 4800], [9000, 8500, 9500]]) 
```

```python
count=np.sum(steps)
count
```

```python
mean=np.mean(steps)
mean
```

```python
median=np.median(steps)
median
```

```python
maximum=np.max(steps)
minimum=np.min(steps)
maximum
```

```python
minimum
```

```python
daily_steps=np.sum(steps,axis=0)
daily_steps
```

```python
user_steps=np.sum(steps,axis=1)
user_steps
```

```python
position = np.argmax(steps)
print(position)
```

```python
position = np.unravel_index(np.argmax(steps), steps.shape)

print(position)
```

```python
X = np.array([ [6, 70, 3], [5, 50, 6], [8, 80, 2], [4, 30, 8] ]) 
```

```python
y = np.array([40, 65, 30, 85]) 
```

```python
X.shape
```

```python
X.ndim
```

```python
XT = X.T
print(XT)
```

```python
XTX = XT @ X
print(XTX)
```

(XTX)−1

```python
inverse = np.linalg.inv(XT @ X)
inverse
```

β=(XTX)−1XTy

```python

beta = np.linalg.inv(X.T @ X) @ X.T @ y

print(beta)
```

```python
new_user = np.array([5, 40, 7])

predicte = new_user @ beta
predicte
```
