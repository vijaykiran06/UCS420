```python
n = int(input("Enter a number: "))
s = 0
for i in range(1, n + 1):
    if i % 7 == 0 and i % 9 == 0:
        s = s + i

print("Sum of numbers divisible by both 7 and 9 from 1 to", n, " --> ", s)
```

assingment 5.3

```python
n = int(input("Enter a number: "))
s = 0
for num in range(2, n + 1):
    is_prime = True
    for i in range(2, num // 2 + 1):
        if num % i == 0:
            is_prime = False
            break
    if is_prime:
        s = s + num

print("Sum of prime numbers from 1 to", n, " --> ", s)
```

assingment 6.1

```python
def AddOdd(n):
    s = 0
    for i in range(1, n + 1):
        if i % 2 != 0:
            s = s + i
    return s

n = int(input("Enter a number: "))
print("Sum of odd numbers from 1 to", n, " --> ", AddOdd(n))
```

assingemnt 6.2

```python
def IsPrime(num):
    if num < 2:
        return 0
    for i in range(2, num // 2 + 1):
        if num % i == 0:
            return 0
    return 1
def AddPrime(n):
    s = 0
    for i in range(2, n + 1):
        if IsPrime(i):
            s = s + i
    return s

n = int(input("Enter a number: "))
print("Sum of prime numbers from 1 to", n, " --> ", AddPrime(n))
```
